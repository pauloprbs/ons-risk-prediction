```python
%%writefile source_lstm/train_lstm.py
import argparse
import os
import pandas as pd
import numpy as np
import joblib
import sys
import traceback

# TensorFlow e Keras
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import LSTM, Dense, Dropout, BatchNormalization, Bidirectional
from tensorflow.keras.callbacks import EarlyStopping
import keras_tuner as kt

# Scikit-learn
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import classification_report

# Imbalanced-learn
from imblearn.over_sampling import RandomOverSampler

# Definição de features
features_hidrica = ['ear_percentual_seco', 'ena_percentual_mlt_seco']
features_clima = ['ghi', 'temp2m_c', 'precipitacao_mm']
features_calendario = ['mes', 'dia_da_semana', 'dia_do_ano']
features_avancadas = [
    'carga_media_7d', 'carga_std_7d', 'geracao_media_7d', 'ear_ontem',
    'ear_diff_3d', 'margem_oferta_demanda', 'pressao_demanda_ear',
    'precip_acumulada_14d', 'precip_acumulada_30d'
]
features_adicionais = ['cmo_semanal_seco', 'disponibilidade_total_diaria_go']
features_geracao = [
    'geracao_total_diaria_go', 'geracao_fotovoltaica_diaria',
    'geracao_hidroelétrica_diaria', 'geracao_térmica_diaria'
]
features_rede = ['total_mwh_restrito_go', 'saldo_intercambio_seco']

def get_features_for_scenario(cenario):
    if cenario == 'pos_2022':
        return (features_geracao + features_rede + features_hidrica +
                features_adicionais + features_clima + features_calendario +
                features_avancadas + ['programada'])
    elif cenario == 'pos_2020':
        return (features_hidrica + features_adicionais + features_clima +
                features_calendario + features_avancadas + ['programada'])
    elif cenario in ['pos_2017', 'pos_2015']:
        return (features_hidrica + features_adicionais + features_clima +
                features_calendario + features_avancadas)
    elif cenario in ['pos_2013', 'pos_2010']:
        return (features_hidrica + features_clima + features_calendario +
                features_avancadas + ['cmo_semanal_seco', 'indicador_ccal_mensal'])
    else:
        raise ValueError(f"Cenário '{cenario}' não reconhecido")

def create_sequences(X, y, time_steps=14):
    Xs, ys = [], []
    for i in range(len(X) - time_steps):
        Xs.append(X[i:i + time_steps])
        ys.append(y[i + time_steps])
    return np.array(Xs), np.array(ys)

def build_model(hp, input_shape):
    model = Sequential()
    
    hp_units_1 = hp.Int('units_1', min_value=64, max_value=256, step=64)
    hp_dropout_1 = hp.Float('dropout_1', min_value=0.2, max_value=0.5, step=0.1)
    model.add(Bidirectional(LSTM(units=hp_units_1, return_sequences=True), input_shape=input_shape))
    model.add(Dropout(hp_dropout_1))
    
    hp_units_2 = hp.Int('units_2', min_value=32, max_value=128, step=32)
    hp_dropout_2 = hp.Float('dropout_2', min_value=0.2, max_value=0.5, step=0.1)
    model.add(Bidirectional(LSTM(units=hp_units_2, return_sequences=True)))
    model.add(Dropout(hp_dropout_2))
    
    hp_units_3 = hp.Int('units_3', min_value=16, max_value=64, step=16)
    hp_dropout_3 = hp.Float('dropout_3', min_value=0.2, max_value=0.5, step=0.1)
    model.add(LSTM(units=hp_units_3))
    model.add(Dropout(hp_dropout_3))
    
    hp_dense_1 = hp.Int('dense_1', min_value=32, max_value=128, step=32)
    hp_dense_2 = hp.Int('dense_2', min_value=16, max_value=64, step=16)
    model.add(Dense(units=hp_dense_1, activation='relu'))
    model.add(BatchNormalization())
    model.add(Dense(units=hp_dense_2, activation='relu'))
    
    model.add(Dense(3, activation='softmax'))
    
    hp_learning_rate = hp.Choice('learning_rate', values=[1e-2, 1e-3, 1e-4])
    model.compile(optimizer=tf.keras.optimizers.Adam(learning_rate=hp_learning_rate),
                  loss='sparse_categorical_crossentropy',
                  metrics=['accuracy'])
    return model

if __name__ == '__main__':
    parser = argparse.ArgumentParser()
    
    parser.add_argument('--model-dir', '--model_dir', type=str, default=os.environ.get('SM_MODEL_DIR'), dest='model_dir')
    parser.add_argument('--train', type=str, default=os.environ.get('SM_CHANNEL_TRAIN'))
    
    parser.add_argument('--cenario_escolhido', type=str, required=True)
    parser.add_argument('--time_steps', type=int, default=14)
    parser.add_argument('--test_size', type=float, default=0.2)
    parser.add_argument('--max_trials', type=int, default=10)
    parser.add_argument('--executions_per_trial', type=int, default=2)
    parser.add_argument('--epochs', type=int, default=50)
    parser.add_argument('--patience', type=int, default=10)
    
    args = parser.parse_args()
    
    print("--- Iniciando Treinamento LSTM ---")
    print(f"Parâmetros: {vars(args)}")
    
    try:
        input_files_path = args.train
        all_files = [os.path.join(input_files_path, f) for f in os.listdir(input_files_path) if f.endswith('.parquet')]
        
        if not all_files:
            raise ValueError(f"Nenhum arquivo .parquet encontrado em {input_files_path}")
        
        print(f"\nArquivos encontrados: {len(all_files)}")
        for f in all_files:
            print(f"  - {os.path.basename(f)}")
        
        # Ler todos os arquivos
        list_of_dfs = [pd.read_parquet(f) for f in all_files]
        df_completo = pd.concat(list_of_dfs, ignore_index=True)
        
        print(f"\nShape do DataFrame completo: {df_completo.shape}")
        print(f"Colunas disponíveis ({len(df_completo.columns)}): {df_completo.columns.tolist()[:10]}...")
        
        # CORREÇÃO: Detectar corretamente a coluna alvo
        target_column = None
        
        if 'nivel_risco' in df_completo.columns:
            target_column = 'nivel_risco'
            print(f"\n✓ Coluna alvo encontrada: '{target_column}'")
        else:
            target_column = df_completo.columns[-1]
            print(f"\n⚠ Coluna 'nivel_risco' não encontrada. Usando última coluna: '{target_column}'")
        
        # Separar features e target
        features_para_treino = get_features_for_scenario(args.cenario_escolhido)
        
        features_disponiveis = [col for col in features_para_treino if col in df_completo.columns and col != target_column]
        
        if not features_disponiveis:
            print("\n⚠ AVISO: Nenhuma feature do cenário encontrada. Usando todas as colunas exceto a última.")
            features_disponiveis = [col for col in df_completo.columns if col != target_column]
        
        print(f"\nFeatures selecionadas ({len(features_disponiveis)}): {features_disponiveis[:10]}...")
        
        X = df_completo[features_disponiveis]
        y = df_completo[target_column]
        
        print(f"\nShape de X: {X.shape}")
        print(f"Shape de y: {y.shape}")
        print(f"Valores únicos em y: {y.unique()[:10]}")
        
        local_model_dir = '/opt/ml/model'
        os.makedirs(local_model_dir, exist_ok=True)
        
        # Encoding da variável alvo
        if y.dtype == 'object' or y.dtype.name == 'category':
            print("\nTarget é categórico. Aplicando mapeamento...")
            mapeamento_risco = {'baixo': 0, 'medio': 1, 'alto': 2}
            y_encoded = y.map(mapeamento_risco)
            
            if y_encoded.isna().any():
                print("⚠ AVISO: Valores não mapeados encontrados. Usando LabelEncoder...")
                from sklearn.preprocessing import LabelEncoder
                le = LabelEncoder()
                y_encoded = pd.Series(le.fit_transform(y), index=y.index)
                joblib.dump(le, os.path.join(local_model_dir, 'label_encoder.joblib'))
        else:
            print("\nTarget é numérico. Aplicando binning por quantis...")
            q1 = y.quantile(0.33)
            q2 = y.quantile(0.67)
            y_encoded = pd.Series(index=y.index, dtype=int)
            y_encoded[y <= q1] = 0
            y_encoded[(y > q1) & (y <= q2)] = 1
            y_encoded[y > q2] = 2
            joblib.dump({'q1': float(q1), 'q2': float(q2)}, os.path.join(local_model_dir, 'risk_thresholds.joblib'))
        
        y_encoded = y_encoded.astype(int)
        print(f"Distribuição das classes: {y_encoded.value_counts().to_dict()}")
        
        # Split temporal
        split_index = int(len(X) * (1 - args.test_size))
        X_train_full, y_train_full = X[:split_index], y_encoded[:split_index]
        X_test, y_test = X[split_index:], y_encoded[split_index:]
        
        print(f"\nTrain size: {len(X_train_full)}, Test size: {len(X_test)}")
        
        # Limpeza de dados
        X_train_full = X_train_full.dropna(axis=1, how='all')
        cols_to_keep = X_train_full.var()[X_train_full.var() > 0].index
        X_train_full, X_test = X_train_full[cols_to_keep], X_test[cols_to_keep]
        X_train_full.fillna(X_train_full.median(), inplace=True)
        X_test.fillna(X_test.median(), inplace=True)
        
        print(f"Features após limpeza: {X_train_full.shape[1]}") # Deve imprimir 22

        # Salvar a lista final de features (cols_to_keep) ANTES de escalar
        final_feature_names_lstm = cols_to_keep.tolist() # Convertendo Index para lista
        print(f"Capturando {len(final_feature_names_lstm)} nomes de features finais para LSTM.")
        features_path = os.path.join(local_model_dir, 'feature_names_lstm.joblib')
        print(f"Salvando features do LSTM em: {features_path}")
        joblib.dump(final_feature_names_lstm, features_path)
        
        # Normalização
        scaler = StandardScaler()
        X_train_scaled = scaler.fit_transform(X_train_full)
        X_test_scaled = scaler.transform(X_test)
        
        # Criar sequências
        print(f"\nCriando sequências com time_steps={args.time_steps}...")
        X_train_seq, y_train_seq = create_sequences(X_train_scaled, y_train_full.values, args.time_steps)
        X_test_seq, y_test_seq = create_sequences(X_test_scaled, y_test.values, args.time_steps)
        
        print(f"Shape das sequências de treino: {X_train_seq.shape}")
        print(f"Shape das sequências de teste: {X_test_seq.shape}")
        
        # Oversampling
        print("\nAplicando RandomOverSampler...")
        ros = RandomOverSampler(random_state=42)
        # Reshape para 2D antes do ROS
        X_train_seq_2d = X_train_seq.reshape(X_train_seq.shape[0], -1) 
        X_res_2d, y_res = ros.fit_resample(X_train_seq_2d, y_train_seq)
        # Reshape de volta para 3D
        X_res = X_res_2d.reshape(-1, X_train_seq.shape[1], X_train_seq.shape[2]) 
        
        print(f"\nShape após oversampling: {X_res.shape}")
        print(f"Distribuição balanceada: {pd.Series(y_res).value_counts().to_dict()}")
        
        # Keras Tuner - definir wrapper antes
        def build_model_wrapper(hp):
            return build_model(hp, (X_res.shape[1], X_res.shape[2]))
        
        # Keras Tuner - usar diretório local temporário
        tuner_dir = '/tmp/keras_tuner'
        os.makedirs(tuner_dir, exist_ok=True)
        
        print(f"\nIniciando Keras Tuner com max_trials={args.max_trials}...")
        tuner = kt.RandomSearch(
            build_model_wrapper,
            objective=kt.Objective("val_accuracy", direction="max"),
            max_trials=args.max_trials,
            executions_per_trial=args.executions_per_trial,
            directory=tuner_dir,
            project_name=f'lstm_{args.cenario_escolhido}',
            overwrite=True
        )
        
        early_stopping = EarlyStopping(monitor='val_loss', patience=args.patience, restore_best_weights=True)
        tuner.search(X_res, y_res, epochs=args.epochs, validation_split=0.2, callbacks=[early_stopping], verbose=1)
        
        # Melhor modelo
        best_model = tuner.get_best_models(1)[0]
        print("\n✓ Melhor modelo obtido!")
        
        # Avaliação
        y_pred_proba = best_model.predict(X_test_seq, verbose=0)
        y_pred = np.argmax(y_pred_proba, axis=1)
        
        print("\n--- Classification Report ---")
        print(classification_report(y_test_seq, y_pred, target_names=['baixo', 'medio', 'alto'], zero_division=0))
        
        # Salvar modelo e scaler no diretório local
        model_path = os.path.join(local_model_dir, f'lstm_model_{args.cenario_escolhido}.h5')
        scaler_path = os.path.join(local_model_dir, f'scaler_lstm_{args.cenario_escolhido}.joblib')
        
        best_model.save(model_path)
        joblib.dump(scaler, scaler_path)
        
        # Salvar informações adicionais (metadata já contém as features corretas)
        metadata = {
            'cenario': args.cenario_escolhido,
            'time_steps': args.time_steps,
            'n_features': X_train_seq.shape[2], # Deve ser 22
            'classes': ['baixo', 'medio', 'alto'],
            'feature_names': final_feature_names_lstm # Salva a lista de 22 features
        }
        joblib.dump(metadata, os.path.join(local_model_dir, 'model_metadata.joblib'))
        
        print(f"\n✓ Modelo salvo em: {model_path}")
        print(f"✓ Scaler salvo em: {scaler_path}")
        print(f"✓ Features salvas em: {features_path}") # Adicionado print
        print(f"✓ Metadata salvo em: {os.path.join(local_model_dir, 'model_metadata.joblib')}")
        print("\n--- Treinamento Finalizado com Sucesso ---")
        
    except Exception as e:
        print(f"\n❌ ERRO: {e}\n")
        traceback.print_exc()
        sys.exit(255)
```

```python
from sagemaker.tensorflow import TensorFlow

# === PAINEL DE CONTROLE ===
CENARIO_ESCOLHIDO = 'pos_2010'  # Opções: pos_2022, pos_2020, pos_2017, pos_2015, pos_2013, pos_2010
INSTANCE_TYPE = 'ml.m5.xlarge'
# ===========================

# Hiperparâmetros
hyperparameters = {
    'cenario_escolhido': CENARIO_ESCOLHIDO,
    'time_steps': 14,
    'test_size': 0.2,
    'max_trials': 10,
    'executions_per_trial': 2,
    'epochs': 50,
    'patience': 10
}

# Criar estimador TensorFlow
lstm_estimator = TensorFlow(
    entry_point='train_lstm.py',
    source_dir=source_dir,
    role=role,
    instance_count=1,
    instance_type=INSTANCE_TYPE,
    framework_version='2.13',
    py_version='py310',
    hyperparameters=hyperparameters,
    output_path=s3_model_output_path,
    sagemaker_session=sess,
    max_run=7200,  # 2 horas max
)

# Definir input de dados
train_input = sagemaker.inputs.TrainingInput(
    s3_data=s3_input_data_path,
    distribution='FullyReplicated',
    content_type='parquet',
    s3_data_type='S3Prefix'
)

# Nome único do job
timestamp = time.strftime("%Y%m%d-%H%M%S", time.gmtime())
job_name = f'lstm-ons-risk-{CENARIO_ESCOLHIDO.replace("_", "-")}-{timestamp}'

print(f"Iniciando Job de Treinamento LSTM: {job_name}")
print(f"Usando dados de: {s3_input_data_path}")
print(f"Modelo será salvo em: {s3_model_output_path}")
print(f"Hiperparâmetros: {hyperparameters}")
print(f"\n⚠️  ATENÇÃO: Treinamento LSTM pode levar 30-60 minutos com Keras Tuner!")

# Lançar treinamento
lstm_estimator.fit({'train': train_input}, job_name=job_name, wait=True)

# Recuperar caminho do modelo
lstm_model_s3_uri = lstm_estimator.model_data
print(f"\nJob de Treinamento {job_name} concluído!")
print(f"Modelo salvo em: {lstm_model_s3_uri}")
```
